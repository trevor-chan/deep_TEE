```python
fig = go.Figure()

fig.update_layout(
    plot_bgcolor='rgba(0,0,0,0)',
    paper_bgcolor='rgba(0,0,0,0)'
)

for i, value in enumerate(values):
    fig.add_trace(go.Bar(
        y=voting_scheme,
        x=value,
        name=labels[i],
        orientation='h',
        marker=dict(
            color=colors[labels[i]],
            line=dict(color=colors[labels[i]], width=3)
        )
    ))

fig.update_layout(barmode='stack')
fig.show()
```
